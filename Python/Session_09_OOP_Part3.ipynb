### Aggregation(Has-A relationship)  
One class owns the other class.

### This is aggregation

```python
class Customer:

  def __init__(self,name,gender,address):
    self.name=name
    self.gender=gender # name and gender are simple
    self.address=address # address is complex entity, having house number, zip, area ,city etc.

  def print_statement(self):
    print(self.address.get_city(),self.address.pin,self.address.state)
# we are making getter cause it will throw error we can't access private var.

  def edit_profile(self,new_name,new_city,new_pin,new_state):
    self.name=new_name
    self.address.edit_address(new_city,new_pin,new_state) # edit address is called inside address class.

class Address:

  def __init__(self,city,pin,state):
    self.__city=city
    self.pin=pin
    self.state=state

  def get_city(self): # getter
    return self.__city

  def edit_address(self,new_city,new_pin,new_state): # setter
    self.__city=new_city
    self.pin=new_pin
    self.state=new_state

add1=Address('delhi',122011,'UP') # this is object of address class
cust=Customer("bibek",'male',add1) # cust obj is made of Customer class and in address, add1 is passed.

cust.print_statement()
cust.edit_profile('sumit','jhapa',1234,'koshi') # name is going to customer class, but address is changed with address class.

cust.print_statement()
```

### Super Keyword(IMP)  
We can use super to call parent constructor/methods from child class if we want.    We can't use parent attribute using super.

*We can't use super keyword outside of class. It is mainly used inside of class*

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

```python
# can super access parent ka data?

class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka BRAND DATA
        print(super().brand) #we can't get attributes of parent from super()
s=SmartPhone(20000, "Apple", 13)

s.buy()
```

```python
# super -> constuctor
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera) # calling parent constructor
        self.os = os
        self.ram = ram
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", 12, "Android", 2)

print(s.os)
print(s.brand)
print(s.os)
```

##### Inheritance in summary

- A class can inherit from another class.

- Inheritance improves code reuse

- Constructor, attributes, methods get inherited to the child class

- The parent has no access to the child class

- Private properties of parent are not accessible directly in child class

- Child class can override the attributes or methods. This is called method overriding

- super() is an inbuilt function which is used to invoke the parent class methods and constructor

```python
class Parent:

    def __init__(self,num):
      self.__num=num

    def get_num(self):
      return self.__num

class Child(Parent):

    def __init__(self,num,val):
      super().__init__(num)
      self.__val=val

    def get_val(self):
      return self.__val

son=Child(100,200)
print(son.get_num())
print(son.get_val())
```

### Types of Inheritance

- Single Inheritance
- Multilevel Inheritance
- Hierarchical Inheritance
- Multiple Inheritance(Diamond Problem)
- Hybrid Inheritance

```python
# single inheritance
class Phone: #parent
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone): # child
    pass

SmartPhone(1000,"Apple","13px").buy()
```

```python
# multilevel
#IT can inherit from all the above parent class.
class Product: # grandfather
    def review(self):
        print ("Product customer review")

class Phone(Product): # father
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone): # son
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy() # inherit fathr clss
s.review() #  inherit grandfather clss
```

```python
# Hierarchical
#ONe parent and multiple children, like father has two or more than two kids.
class Phone: # one parent
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone): #one child
    pass

class FeaturePhone(Phone): # another child
    pass

SmartPhone(1000,"Apple","13px").buy()
FeaturePhone(10,"Lava","1px").buy()
```

```python
# Multiple
#child class can inherit from mother and father classs.
class Phone: # father class
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product: # mother class
    def review(self):
        print ("Customer review")

class SmartPhone(Phone, Product): # this is child class
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
#This is not applicable in Java due to ambuigity.
```

```python
#hybrid
#- can consists of combination of all the above.

# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product): # Phone is passed at first so It'll run Phone class buy method.
    pass
#This concept is called method resolution order (MRO).
s=SmartPhone(20000, "Apple", 12)

s.buy()
```

# Polymorphism  
The same interface/method call can behave differently depending on the object.

- ### Method Overriding  
*Method overriding happens when a child class has a method with the same name as a method in its parent class, and the child provides its own implementation. When obj from child class its made, its calls its own method if both parent and child has same method.*
- ### Method Overloading  
*Method overloading means having multiple methods with the same name but different parameters, so the method behaves differently depending on the arguments.*
- ### Operator Overloading  
*Operator overloading means giving an operator such as +, -, >, ==, etc. a specific meaning for objects of your own class.*

```python
#Methhod overloading
#This code don't work in python but works in java.
class Shape:

  def area(self,radius):
    return 3.14*radius*radius

  def area(self,l,b):
    return l*b

s=Shape()
s.area(3)
s.area(3,4)
```

```python
# We can achieve this  in python by using its special feature default parameter

class Shape:

  def area(self,a,b=0):
    if b==0:
      return 3.14*a*a
    else:
      return a*b

s=Shape()
print(s.area(3)) # if only one argument radius works
s.area(3,4) # if two arguments, l and b works.
```

# Abstraction (Hidden)  
### Least used oop feature.

For eg: a laptop has many technical things inside it like cpu, hardware, ram, graphics card etc. but one person who uses laptop doesn't care about it as long he knows how to use a laptop.  
So the core component of laptops are hidden and not shown to outer world.

**Abstraction**: Show only what the user needs to use, and hide the internal implementation/details.  


```python
#abstract class is that class which has atleast one abstract method
# we can't make objects from abstract class.
from abc import ABC,abstractmethod
class BankApp(ABC):

  def database(self):
    print("connected to database")

  @abstractmethod # decorater telling it is abstract class.
  def security(self):
    pass
```

```python
class MobileApp(BankApp):
  def mobile_login(self):
    print('login into mobile')


  def security(self):
      print("mobile security")
```

```python
mob=MobileApp()
```

```python
mob.database()
```

**Think about it as a contract**

```python
from abc import ABC,abstractmethod
class Animal(ABC):

    @abstractmethod # this is a rule.
    def sound(self):
        pass


# Any child class of Animal MUST provide a sound() method.
```

```python
class Dog(Animal):

    def sound(self):
        print("Bark")
d1=Dog()
d1.sound() # this works because child class Dog has method sound which is a rule defined in parent class.
```

```python
# But if there is a class without sound method. It doesn't work.
class Cat(Animal):

    def eat(self):
      return "Cat eat food"

d2=Cat() # This gives error because the child class doesn't have sound method which parent class tells it should must have.
```

# Why do we need abstraction?

Imagine you're building a payment system.  
We don't want the Payment class to explain how every payment works. Different payment systems have different internal processes.

So we only define the thing that every payment system must provide:

You might say:



```python
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass
```

You don't care how payment happens at the general level.

You just establish:

Every payment method must have a pay() method.

Then:

```python
class Esewa(Payment):

    def pay(self, amount):
        print("Paying through eSewa")
```

```python
class Khalti(Payment):

    def pay(self, amount):
        print("Paying through Khalti")
```

```python
e = Esewa()
e.pay(1000)
k=Khalti()
k.pay(5000)
```
