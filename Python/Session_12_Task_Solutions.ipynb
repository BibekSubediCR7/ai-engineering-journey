### `Q1:` Write `Person` Class as given below and then display it's namespace.



```
Class Name - Person

Attributes:
name - public
state - public
city - private
age - private

Methods:
address - public
It give address of the person as "<name>, <city>, <state>"
```

````python
#Write your code here
class Person:
    def __init__(self, name, state):
        self.name = name
        self.state = state
        self.__city = None
        self.__age = None
    def set_city(self, city):
        self.__city = city
    def get_city(self):
        return self.__city
    def set_age(self, age):
        self.__age = age
    def get_age(self):
        return self.__age
    def address(self):
        return f'{self.name}, {self.__city}, {self.state}'

for i in Person.__dict__:
    print(i)
````

### Explaining Q1: Understanding Class Namespaces in Python

Alright, let's dive into the first question! We're creating a `Person` class, which is like a blueprint for creating individual person objects. Think of it as defining what a 'person' means in our program.

*   **`class Person:`**: This is where we declare our new class, simply named `Person`.

*   **`def __init__(self, name, state):`**: This is a very special method called the **constructor**. It's the first thing that runs whenever you create a *new* `Person` object.
    *   `self` is a convention in Python; it always refers to the specific object (instance) that's being created right now.
    *   `name` and `state` are pieces of information we *pass in* when we create a person (e.g., `Person("Raj", "West Bengal")`).
    *   `self.name = name` and `self.state = state`: These lines store the `name` and `state` you passed in directly as *public* attributes of our person object. Anyone can easily access these.
    *   `self.__city = None` and `self.__age = None`: See those double underscores (`__`)? That's Python's way of saying, "Hey, these are *private*!" It's a strong hint that these attributes (`city` and `age`) should primarily be handled by methods inside the class, not directly from outside. We initialize them to `None` because we don't have values for them yet.

*   **`set_city`, `get_city`, `set_age`, `get_age` methods**: These are what we call **setter** and **getter** methods. They act as controlled doorways to interact with our private `__city` and `__age` attributes. Instead of directly saying `p.__city = "Kolkata"` (which is generally bad practice for private attributes), you'd use `p.set_city("Kolkata")`. This is a core concept in object-oriented programming called **encapsulation** – bundling data and the methods that operate on that data within one unit.

*   **`def address(self):`**: This is a public method that simply builds a nice address string using the person's `name`, `city`, and `state`.

*   **`for i in Person.__dict__:`**: Now, this is the crucial part for understanding **namespaces**! `Person.__dict__` isn't about an *individual person*; it's about the `Person` *class itself*. It holds all the names (like `__init__`, `set_city`, `address`, `__module__`, etc.) that are *directly defined within the class blueprint*. When you print `i`, you're seeing these names, not the specific values of `Raj` or `Kolkata`. This is the class's own dictionary of what it contains.

### Explaining Q2: Object/Instance Namespaces – What an Individual Knows

Building on our `Person` class from Q1, now we're looking at what happens when we create an *actual person* from that blueprint. This is about the **namespace of an object**, also known as an **instance**.

*   **`p = Person("Raj", "West Bengal")`**: Here, we're bringing our `Person` blueprint to life! `p` is now an *instance* of the `Person` class. When this line runs, our `__init__` method (from Q1) is automatically called, and `p` gets its `name` attribute set to "Raj" and `state` to "West Bengal".

*   **`p.set_city("Kolkata")`** and **`p.set_age(40)`**: We're using our trusty setter methods to assign values to the *private* `__city` and `__age` attributes for *this specific person*, `p`.

*   **`print("Address: ", p.address())`**: We call the `address` method on our `p` object to see its formatted address.

*   **`for i in p.__dict__:`**: This is the key line for this question! Unlike `Person.__dict__` (which showed what the *class* defines), `p.__dict__` shows us what's unique to *this particular object* `p`.
    *   The `__dict__` of an instance (like `p`) contains all the attributes that are specific to *that* instance. These are usually the attributes that were created or modified *after* the object was initialized.
    *   When you print `i`, you'll see `name`, `state`, `_Person__city`, and `_Person__age`.
        *   `name` and `state` are straightforward, as we assigned them directly.
        *   But wait, why `_Person__city` and `_Person__age` instead of `__city` and `__age`? This is Python's **name mangling** at work! When you prefix an attribute with double underscores (`__`) in a class, Python automatically changes its name internally to include the class name (e.g., `_Person__city`). This helps prevent naming conflicts, especially in more complex inheritance scenarios, and reinforces the idea that these are "private" attributes meant for internal use by the class methods. It's a clever way Python helps you manage attribute visibility!

### `Q3:` Write a recursive program to to calculate `gcd` and print no. of function calls taken to find the solution.
```
gcd(5,10) -> result in 5 as gcd and function call 3
```

````python
#Write your code here
# O(log(min(a,b)))

counter = 0
def gcd(a,b):
    global counter
    counter += 1
    if b == 0:
        return a
    else:
        return gcd(b, a%b)
print(gcd(10, 5), counter)
````

### Explaining Q3: Recursive GCD with a Call Counter

Let's break down this `gcd` (Greatest Common Divisor) function, which uses a powerful technique called **recursion** and also counts how many times it's called!.

*   **`counter = 0`**: We start by setting up a `counter` variable outside our function. We want this `counter` to keep track of *all* calls across all the recursive steps, so it needs to be a **global** variable.

*   **`def gcd(a, b):`**: This defines our `gcd` function, which takes two numbers, `a` and `b`.

*   **`global counter`**: Inside the function, this line is crucial! It tells Python, "When I say `counter` here, I mean the `counter` variable that's defined *globally*, not a new, separate one just for this function call." This allows us to modify the single `counter` variable.

*   **`counter += 1`**: Every single time `gcd` is called (whether it's the initial call or a recursive call), we increment this `counter`. This is how we track the number of function calls.

*   **`if b == 0:`**: This is our **base case**. In recursion, a base case is the condition that tells the function when to stop calling itself. According to the Euclidean algorithm (a famous way to find GCD), if one number (`b` in this case) is 0, then the other number (`a`) is the GCD. So, we `return a`, and the recursion unwinds.

*   **`else: return gcd(b, a % b)`**: This is the **recursive step**. If `b` isn't 0, we don't know the GCD yet. So, we call `gcd` *again*, but with new arguments: `b` becomes the first argument, and `a % b` (the remainder of `a` divided by `b`) becomes the second. This step moves us closer to the base case because `a % b` will always be smaller than `b`.

*   **`print(gcd(10, 5), counter)`**: Finally, we call `gcd(10, 5)`. The function will recursively call itself until `b` becomes 0, returning the GCD. Once the `gcd` function is completely finished and has returned its final result, we print that GCD *and* the `counter`'s final value, showing how many calls it took.

**Let's trace `gcd(10, 5)`:**
1.  **`gcd(10, 5)`**: `counter` becomes 1. `b` (5) is not 0. Calls `gcd(5, 10 % 5)`, which simplifies to `gcd(5, 0)`.
2.  **`gcd(5, 0)`**: `counter` becomes 2. `b` (0) *is* 0. It hits the base case and returns `5`.
3.  The first `gcd(10, 5)` call receives `5` from its recursive call and returns `5`.

So, the result is `5` for the GCD, and the `counter` correctly shows `2` function calls!

### `Q4:` Create MyEnumerate class,
Create your own `MyEnumerate` class such that someone can use it instead of enumerate. It will need to return a `tuple` with each iteration, with the first element in the tuple being the `index` (starting with 0) and the second element being the `current element` from the underlying data structure. Trying to use `MyEnumerate` with a noniterable argument will result in an error.

```
for index, letter in MyEnumerate('abc'):
    print(f'{index} : {letter}')
```

Output:
```
0 : a
1 : b
2 : c
```

````python
#Write your code here

class MyEnumerate:
    def __init__(self, data):
        self.data = data
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration
        value = (self.index, self.data[self.index])
        self.index += 1
        return value
for index, letter in MyEnumerate('abc'):
    print(f'{index} : {letter}')
````

### Explaining Q4: Building Your Own `MyEnumerate` Class

Have you ever used Python's built-in `enumerate` function to get both the index and the item while looping? It's super handy! In this question, we're building our own version called `MyEnumerate` to understand how Python's **iterators** work under the hood.

For an object to be *iterable* (meaning you can use it in a `for...in` loop), it needs to implement two special methods: `__iter__` and `__next__`.

*   **`class MyEnumerate:`**: This just defines our new class.

*   **`def __init__(self, data):`**: Our constructor. It takes the `data` (like a string, list, etc.) that we want to enumerate.
    *   `self.data = data`: We store this data.
    *   `self.index = 0`: We initialize an `index` to 0. This will keep track of our current position as we iterate.

*   **`def __iter__(self):`**: This method is called whenever Python needs to get an *iterator* from an object (for example, when a `for` loop starts). For an iterator, `__iter__` typically just returns `self`, meaning the object *itself* is its own iterator.

*   **`def __next__(self):`**: This is the heart of our iterator! The `for` loop repeatedly calls `__next__` to get the next item:
    *   **`if self.index >= len(self.data):`**: This is our **stopping condition**. If our `index` has gone past the length of our `data`, it means we've run out of items. To signal the `for` loop to stop, we `raise StopIteration`. This is the standard Python way to end an iteration.
    *   **`value = (self.index, self.data[self.index])`**: If we still have items, we create a `tuple`. The first element is our current `index`, and the second is the actual `item` from our `data` at that `index`.
    *   **`self.index += 1`**: We increment our `index` so we're ready for the next item in the next call to `__next__`.
    *   **`return value`**: We return the `(index, item)` tuple.

*   **`for index, letter in MyEnumerate('abc'):`**: This line shows it all in action!
    1.  `MyEnumerate('abc')` creates an instance of our class.
    2.  The `for` loop implicitly calls `__iter__` on this instance to get the iterator.
    3.  Then, it repeatedly calls `__next__`. Each time `__next__` returns a tuple, the `for` loop cleverly unpacks it into `index` and `letter` for us to print. This continues until `StopIteration` is raised.

### `Q5:` Iterate in circle
Define a class, `Circle`, that takes two arguments when defined: a sequence and a number. The idea is that the object will then return elements the defined number of times. If the number is greater than the number of elements, then the sequence  repeats as necessary. You can define an another class used as a helper (like I call `CircleIterator`).

```
c = Circle('abc', 5)
d = Circle('abc', 7)
print(list(c))
print(list(d))
```

Output
```
[a, b, c, a, b]
[a, b, c, a, b, c, a]
```

````python
#Write your code here
class Circle:
    def __init__(self, data, max_iters):
        self.data = data
        self.index = 0
        self.max_iters = max_iters

    def __iter__(self):
        return self
    def __next__(self):
        if self.index >= self.max_iters:
            raise StopIteration
        value = self.data[self.index % len(self.data)]
        self.index += 1
        return value
c = Circle('abc', 5)
d = Circle('abc', 7)
print(list(c))
print(list(d))
````

### Explaining Q5: Iterating in a `Circle` – Repeating Sequences

This `Circle` class is a clever way to iterate through a sequence (like a list or string) multiple times, effectively looping back to the beginning once it reaches the end. It's also an **iterator**, which means it plays by the rules we saw in `MyEnumerate` (`__iter__` and `__next__`).

*   **`class Circle:`**: Our class definition.

*   **`def __init__(self, data, max_iters):`**: The constructor for our `Circle` object:
    *   `self.data = data`: This holds the sequence we want to iterate through (e.g., `'abc'` or `['apple', 'banana']`).
    *   `self.index = 0`: This counter keeps track of how many items we've returned *overall*, not just within a single cycle of the `data`.
    *   `self.max_iters = max_iters`: This is the *total number of items* the user wants to get from our `Circle` iterator before it stops.

*   **`def __iter__(self):`**: Just like with other iterators, this method simply returns `self`.

*   **`def __next__(self):`**: This method is called each time we need the next item from our `Circle`:
    *   **`if self.index >= self.max_iters:`**: This is our **stopping condition**. If our overall `index` has reached or exceeded the `max_iters` (the total number of items requested), we `raise StopIteration` to tell the `for` loop (or `list()` function) that we're done.
    *   **`value = self.data[self.index % len(self.data)]`**: This is the real magic for the "circular" behavior! Let's break it down:
        *   `len(self.data)`: Gets the length of our original sequence (e.g., 3 for `'abc'`).
        *   `self.index % len(self.data)`: The **modulo operator (`%`)** is key here. It gives you the *remainder* of a division.
            *   If `self.index` is 0, 1, or 2, the remainder will be 0, 1, or 2, respectively. So we get `data[0]`, `data[1]`, `data[2]`.
            *   When `self.index` becomes 3, `3 % 3` is 0. So we get `data[0]` again – we've circled back to the beginning!
            *   When `self.index` is 4, `4 % 3` is 1. We get `data[1]` again.
        This effectively maps our ever-increasing `self.index` back to valid indices within our original `data` sequence, creating the circular effect.
    *   **`self.index += 1`**: We increment our overall counter.
    *   **`return value`**: We return the element we just found.

*   **`print(list(c))`** and **`print(list(d))`**: Here, we're creating `Circle` objects and then immediately converting them to lists. When you do `list(c)`, Python repeatedly calls `__next__` on `c` until it receives `StopIteration`, collecting all the yielded items into a list. This clearly shows how `Circle('abc', 5)` yields `a, b, c, a, b` and `Circle('abc', 7)` yields `a, b, c, a, b, c, a`.

### `Q6:` Generator time elapsed
Write a generator function whose argument must be iterable. With each iteration, the generator will return a two-element tuple. The first element in the tuple will be an integer indicating how many seconds have passed since the previous iteration. The tuple’s second element will be the next item from the passed argument.

Note that the timing should be relative to the previous iteration, not when the
generator was first created or invoked. Thus the timing number in the first iteration
will be 0

```
for t in elapsed_since('abcd'):
    print(t)
    time.sleep(2)
```

Output:
```
(0.0, 'a')
(2.005651817999933, 'b')
(2.0023095009998997, 'c')
(2.001949742000079, 'd')
```
Note: Your output may differ because of diffrent system has different processing configuration.

````python
#Write yor code
import time
def elapsed_since(data):
    last_time = time.perf_counter()
    for item in data:
        current_time = time.perf_counter()
        delta = current_time - last_time
        last_time = current_time
        yield (delta, item)

for t in elapsed_since('abcd'):
    print(t)
    time.sleep(2)
````

### Explaining Q6: Generator for Time Elapsed – Measuring the Gaps

This question introduces **generator functions**, a fantastic Python feature for creating iterators that are efficient with memory. Instead of building an entire list and returning it, generators *yield* one item at a time, pausing their execution and resuming it when the next item is requested. This `elapsed_since` generator will help us measure the time between each step of an iteration.

*   **`import time`**: We need the `time` module for its `perf_counter()` function, which is great for precise timing measurements.

*   **`def elapsed_since(data):`**: This is our generator function. Notice it takes an `iterable` (like a string or list) as input.

*   **`last_time = time.perf_counter()`**: Before we start iterating, we capture the current high-resolution time. This `last_time` will serve as our starting point for the first `delta` calculation and then for subsequent calculations.

*   **`for item in data:`**: We start looping through the `data` that was passed in.

*   **`current_time = time.perf_counter()`**: Inside the loop, for each `item`, we capture the current time again.

*   **`delta = current_time - last_time`**: We calculate the difference between the `current_time` and the `last_time`. This `delta` represents how much time has passed *since the last time we measured* (or since the start, for the very first item).

*   **`last_time = current_time`**: This is a critical step! We immediately update `last_time` to be the `current_time` we just captured. Why? Because when the loop comes around for the *next* `item`, we want to measure the time *from this moment*.

*   **`yield (delta, item)`**: This is the `yield` keyword, making this a generator. Instead of `return` (which would end the function), `yield` *pauses* the function, sends the `(delta, item)` tuple back to whoever called it, and remembers its exact state. When the next item is requested, the function picks up exactly where it left off, right after this `yield` line.
    *   For the *first* iteration, `delta` will be extremely small (close to 0), because `current_time` and `last_time` were recorded almost back-to-back.

*   **`for t in elapsed_since('abcd'):`**: Here's how we use it. We call `elapsed_since('abcd')`, which *doesn't run the whole function* but instead gives us a generator object. The `for` loop then requests items one by one from this generator.

*   **`time.sleep(2)`**: After we print each `(delta, item)` tuple, we *intentionally pause* the program for 2 seconds. When the loop asks the generator for the *next* item, the `delta` calculation will now include this 2-second pause, showing you precisely how much time elapsed since the previous `yield`!

## Decorators

### `Q7:` Write a Python program to make a chain of function decorators (bold, italic, underline etc.) on a given function which prints "hello world"

```
def hello():
    return "hello world"
```

```
bold - wrap string with <b> tag. <b>Str</b>
italic - wrap string with <i> tag. <i>Str</i>
underline- wrap string with <u> tag. <u>Str</u>
```

````python
#Write your code here
def make_bold(func):
    def wrapped():
        return "<b>"+ func() + "</b>"
    return wrapped
def make_italic(func):
    def wrapped():
        return "<i>"+ func() + "</i>"
    return wrapped
def make_underline(func):
    def wrapped():
        return "<u>"+ func() + "</u>"
    return wrapped

@make_bold
@make_italic
@make_underline
def hello():
    return "hello world"

print(hello())
````

### Explaining Q7: Chaining Decorators – Wrapping Functions with Style

Alright, let's talk about **decorators**! These are super cool Python features that allow you to modify or "wrap" functions without actually changing their original code. Think of them as adding an extra layer of functionality around your existing functions. When you apply multiple decorators, they form a **chain**, and understanding the order is key!

*   **`def make_bold(func):`**, **`def make_italic(func):`**, **`def make_underline(func):`**: These are our three decorator functions. Each one takes *another function* (`func`) as an argument. Their job is to return a *new* function that wraps the original one.

*   **`def wrapped():`**: Inside each decorator, we define an **inner function**, often called `wrapped` (or `inner`). This `wrapped` function is what will actually *replace* the original function after decoration.
    *   Inside `wrapped`, we call the `func()` (which is `hello()` in our case) and then add HTML tags (`<b>`, `<i>`, `<u>`) around its result. This modifies what `func` would normally return.
    *   **`return wrapped`**: Each decorator returns its `wrapped` inner function. This is how the decorator effectively replaces the original function with its augmented version.

*   **`@make_bold`
    `@make_italic`
    `@make_underline`
    `def hello():`
        `return "hello world"`**:
    This is where we apply our decorators to the `hello` function. The `@` symbol is syntactic sugar for calling the decorator. Here's how the **chain** works:
    1.  **Bottom-up application**: Python applies decorators from the *bottom up* (closest to the function first). So, `hello` is first decorated by `make_underline`. Conceptually, it's like `hello = make_underline(hello)`.
    2.  This means `make_underline` takes the original `hello` and returns a *new* `wrapped` function that adds `<u>` tags. Let's call this `hello_underlined`.
    3.  Next, `hello_underlined` is decorated by `make_italic`. So, `hello_underlined = make_italic(hello_underlined)`. `make_italic` takes `hello_underlined` and returns *another new* `wrapped` function that adds `<i>` tags around the `<u>` tagged string. Let's call this `hello_italic_underlined`.
    4.  Finally, `hello_italic_underlined` is decorated by `make_bold`. So, `hello_italic_underlined = make_bold(hello_italic_underlined)`. `make_bold` takes `hello_italic_underlined` and returns the *final* `wrapped` function that adds `<b>` tags around everything.

*   **`print(hello())`**:
    When you call `hello()`, you're actually calling the *outermost* `wrapped` function (the one from `make_bold`). This `wrapped` function then calls its inner wrapped function (from `make_italic`), which calls its inner wrapped function (from `make_underline`), which finally calls the *original* `hello()` function. The results are then modified step-by-step from the inside out:
    1.  Original `hello()` returns: `"hello world"`
    2.  `make_underline` adds `<u>`: `"<u>hello world</u>"`
    3.  `make_italic` adds `<i>`: `"<i><u>hello world</u></i>"`
    4.  `make_bold` adds `<b>`: `"<b><i><u>hello world</u></i></b>"`

Decorators are a powerful way to add logging, authentication, performance monitoring, or in this case, text formatting, to your functions without cluttering their main logic!

### `Q8:` Write a decorator called `printer` which causes any decorated function to print their return values. If the return value of a given function is `None`, printer should do nothing.



````python
#Write your code here
from functools import wraps
def printer(func):
    @wraps(func)
    def inner(*args, **kwargs):
        return_value = func(*args, **kwargs)
        if return_value is not None:
            print(return_value)
        return return_value
    return inner

@printer
def hello(string):
    return string

hello("hello")
````

### Explaining Q8: The `printer` Decorator – Smart Return Value Output

This question asks us to create a `printer` decorator. Its job is pretty neat: any function it decorates will automatically print its return value *if* that return value isn't `None`. We'll also see a best practice for writing decorators: `functools.wraps`.

*   **`from functools import wraps`**: This is important for robust decorators! `wraps` is itself a decorator that you use *inside* your own decorator. Its main purpose is to preserve the metadata (like the function's original name `__name__`, its documentation `__doc__`, etc.) of the decorated function. Without `wraps`, if you called `help(hello)` after decorating, it would show you information about the inner `inner` function of the decorator, not your original `hello` function. `wraps` fixes this!

*   **`def printer(func):`**: This is our decorator function. It takes the function to be decorated (`func`) as its argument.

*   **`@wraps(func)`**: We apply `wraps` to our `inner` function. This tells `wraps` to copy the metadata *from* `func` *to* `inner`.

*   **`def inner(*args, **kwargs):`**: This is the actual wrapper function that `printer` will return. When a decorated function is called (e.g., `hello("hi")`), it's this `inner` function that gets executed.
    *   `*args` and `**kwargs` are standard Python syntax for allowing a function to accept *any* number of positional arguments (`*args`) and *any* number of keyword arguments (`**kwargs`). This makes our `printer` decorator super flexible; it can decorate functions with any signature!

*   **`return_value = func(*args, **kwargs)`**: Inside our `inner` wrapper, we call the *original* function (`func`) with all the arguments that were passed to `inner`. We then store whatever `func` returns in `return_value`.

*   **`if return_value is not None:`**: Here's the condition! We only want to print something if the original function actually returned a non-`None` value. If a function doesn't explicitly return anything, it implicitly returns `None`.

*   **`print(return_value)`**: If `return_value` is not `None`, we print it to the console.

*   **`return return_value`**: This is crucial! After potentially printing, the `inner` function *always* returns the `return_value` from the original `func`. This ensures that if another part of your program expects a value back from the decorated function, it still gets it, even though our decorator added a printing step.

*   **`return inner`**: Finally, the `printer` decorator itself returns this `inner` function. This `inner` function effectively replaces the original `func` in our program.

*   **`@printer`
    `def hello(string):`
        `return string`**:
    Here, we decorate `hello`. When `hello("hello")` is called:
    1.  The `inner` function from `printer` is what actually executes.
    2.  `inner` calls the *original* `hello("hello")`, which returns the string `"hello"`.
    3.  Since `"hello"` is not `None`, `inner` prints `"hello"`.
    4.  Finally, `inner` returns `"hello"`.

### `Q9:` Make a decorator which calls a given function twice. You can assume the functions don't return anything important, but they may take arguments.
```
#Lets say given function
def hello(string):
    print(string)

#on calling after specified decorator is inplaced
hello('hello')
```

Output:
```
hello
hello
```

````python
#Write your cod here
def double(func):
    @wraps(func)
    def inner(*args, **kwargs):
        func(*args, **kwargs)
        func(*args, **kwargs)
    return inner
@double
def hello(string):
    print(string)

hello("hello")
````

### Explaining Q9: The `double` Decorator – Calling Functions Twice

For this question, we're building a decorator called `double`. Its purpose is straightforward: any function it decorates will automatically execute *twice* every time it's called. This can be handy for things like printing a message twice, or perhaps for some repetitive setup tasks.

*   **`def double(func):`**: This is our decorator function. It takes the original function (`func`, which is `hello` in our example) as its argument.

*   **`@wraps(func)`**: Just like in Q8, we use `functools.wraps` here. This is a good habit for all decorators as it helps preserve the original function's important metadata (`__name__`, `__doc__`, etc.). This means if you tried `help(hello)` after decorating, it would still correctly show you the help for your original `hello` function.

*   **`def inner(*args, **kwargs):`**: This is the wrapper function that our `double` decorator will return. When the decorated function is eventually called, it's this `inner` function that actually runs.
    *   `*args` and `**kwargs` are used to make this decorator versatile. They allow the `inner` function to accept any number of positional arguments and any number of keyword arguments that the *original* `func` might need. So, `double` can decorate functions with different input requirements.

*   **`func(*args, **kwargs)`** (first call):
    *   Inside `inner`, we simply call the *original* `func`, passing along all the arguments that were given to `inner`. This executes the function once.

*   **`func(*args, **kwargs)`** (second call):
    *   Right after the first call, we immediately call the *original* `func` a *second time* with the exact same arguments. This is the core logic of our `double` decorator!

*   **`return inner`**: Our `double` decorator returns this `inner` function. This means that whenever we refer to `hello` after decoration, we're actually referring to this `inner` function.

*   **`@double`
    `def hello(string):`
        `print(string)`**:
    Here, our `hello` function is decorated by `double`. When you call `hello("hello")`:
    1.  The `inner` function (which `double` returned) is what gets executed.
    2.  `inner` first calls the *original* `hello("hello")` once, which prints `"hello"`.
    3.  Then, `inner` calls the *original* `hello("hello")` a *second time*, which prints `"hello"` again.

Since our `hello` function here just prints and doesn't return any important value, the `inner` function doesn't need to explicitly return anything either.

### `Q10:` Write a decorator which doubles the return value of any function. And test that decoratos is working correctly or not using `asert`.

```
add(2,3) -> result in 10. Without decorator it should be 5.
```

````python
# Write your code here
def double(func):
    @wraps(func)
    def inner(*args, **kwargs):
        return func(*args, **kwargs) * 2
    return inner
@double
def add_withDeco(a,b):
    return a+b

def add(a,b):
    return a+b

a=2
b=3
assert add(a,b)*2 == add_withDeco(a,b), "Values are not matching"
print("Values are matching")
````

### Explaining Q10: The `double` Decorator for Return Values – Changing Results

In this final question, we're creating another `double` decorator, but this time its job is to *modify the return value* of the decorated function by multiplying it by 2. We'll also use an `assert` statement to confidently check if our decorator is working correctly.

*   **`def double(func):`**: This is our decorator function, taking the original function (`func`, which is `add_withDeco` in this case) as an argument.

*   **`@wraps(func)`**: We continue using `functools.wraps` here to ensure our decorated function behaves nicely with introspection tools (like `help()`), preserving its original metadata.

*   **`def inner(*args, **kwargs):`**: This is the wrapper function that our `double` decorator will return. It's designed to accept any arguments (`*args`, `**kwargs`) that the decorated function might take.

*   **`return func(*args, **kwargs) * 2`**: This is the core logic that changes the function's output:
    1.  It first calls the *original* `func` with all the arguments it received (`func(*args, **kwargs)`).
    2.  It takes the result of that call.
    3.  It then immediately multiplies that result by `2`.
    4.  Finally, it returns this *doubled* value. This means the external caller will receive the modified result.

*   **`return inner`**: The `double` decorator returns this `inner` function, effectively replacing the original function with this modified version.

*   **`@double`
    `def add_withDeco(a,b):`
        `return a+b`**:
    Here, our `add_withDeco` function is decorated with `double`. When you call `add_withDeco(2, 3)`:
    1.  The `inner` function from the `double` decorator is executed.
    2.  `inner` calls the *original* `add_withDeco(2, 3)`, which calculates `2 + 3 = 5`.
    3.  `inner` then takes this `5`, multiplies it by `2` to get `10`.
    4.  `inner` returns `10`.

*   **`def add(a,b):`
    `return a+b`**: This is a regular `add` function, completely *undecorated*. We're using it as a baseline to compare against our decorated function.

*   **`a=2`, `b=3`**: Simple values for testing.

*   **`assert add(a,b)*2 == add_withDeco(a,b), "Values are not matching"`**: This is a powerful way to test your code! An `assert` statement checks if a condition is `True`. If it's `False`, the program will immediately stop and raise an `AssertionError` with the message you provide. Let's see what it's comparing:
    *   `add(a,b)` will return `5`.
    *   `add(a,b)*2` therefore evaluates to `5 * 2 = 10`.
    *   `add_withDeco(a,b)`, thanks to our decorator, will return `10`.
    *   So, the assertion is effectively checking `10 == 10`, which is `True`! This means our decorator is working exactly as expected.

*   **`print("Values are matching")`**: This line only runs if the `assert` statement passes without an error, confirming the decorator's success!
