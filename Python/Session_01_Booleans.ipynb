```python
# used as conditionals

age = int(input("Enter the age of the person: "))

if (age >= 18) and (age < 60): #if this is true then prints runs.
    print("The person can vote")
elif age >= 60:
    print("Person is a senior citizen")
else:
    print("The person is still a minor, cannot vote")
```

```python
# use of booleans in loops

flag = False

for i in range(1,11):
    print(i)

    if i == 5:
        flag = True

    if flag:
        break
```

```python
# test for equality

# 1 --> True
# 0 --> False

print(bool(0), bool(1))
print(int(False), int(True))
print(float(False), float(True))


print(1 == True)
print(0 != True)
print(not True == 0)
```
