Syntax

real_num + imaginary_num-j

9 + 3j

```python
# complex numbers

com1 = 9 + 3j

print(com1)
```

```python
print(type(com1))
```

```python
# real part

print(com1.real)

# imaginary part

print(com1.imag)
```

```python
# complex()

com2 = complex(real=9, imag=3)

print(com2)
```

```python
com1 == com2
```

```python
# operations

com1 = 10 + 5j

com2 = complex(4 + 2j)

print(com1, com2)
```

```python
# addition

print(com1 + com2)

# subtraction

print(com1 - com2)

# multiplication

print(com1 * com2)
```

```python
# conjugate
print(com1)
print(com1.conjugate())
```

```python
# maginitude

abs(com1)
```
