### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
# Code here
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = round(curr_pop/1.1,1)
```

### Sequence sum

1/1! + 2/2! + 3/3! + ...

```python
# Code here

n = int(input('enter n')) #5

result = 0 #0,1,2
fact = 1 #1,1,2

for i in range(1,n+1): #1,2,3
  fact = fact * i #1,2,6
  result = result + i/fact #1,2,2+3/6

print(result)

```

# Loop Control Statement

- Break: When a condition is met, the loop terminates.  
- Continue: When a condition is met, it skips that condition and goes to next.
- Pass: It is just a placeholder.

```python
for i in range(1,10):
  if i == 5:
    break # it terminates the whole for loop.
  print(i)
```

```python
lower = int(input('enter lower range')) #5
upper = int(input('enter upper range'))#10

for i in range(lower,upper+1):  #(5,11),(6,11),(7,11)
  for j in range(2,i): #(2,5),(3,5),(4,5), (2,6),(2,7), If this loop finishes, else runs.
    if i%j == 0: #5%2F,5%3F,5%4F,6%2T and so on.
      
      break #if i%j is true inner forloop terminates.
  else:
    print(i,end="-") #5
```

```python
# Continue
for i in range(1,10):
  if i == 5:
    continue
  print(i)
```

```python
for i in range(1,10):
  pass
#placeholder , if we can later put value in it.
# for not throwing error while running loops.
```

## Accessing Substrings from a String

## String Slicing

```python
# Slicing - Taking some characters of a word.
s = 'hello world'
print(s[1:5:2]) # start(inclusive, stop(exclusive), step(jump)

print(s[1:]) # This is same as s[1:len(s)]
print(s[:]) # If it has nothing, then starts to end and jump in step size of 1.
```

```python
#Negative indexing.
print(s[6:0:-2]) # start with 6(inclusive) upto 0(exclusive) and reverse jump -2.
#If there is negative indexing, starts with bigger number
```

```python
print(s[::-1]) # reverse the string
#Or same as
print(s[len(s):0:-1])
```

```python
s = 'hello world'
print(s[-1:-6:-1])
```

## Editing and Deleting in Strings
- Strings are immutable which means we can't change the original string.

```python
s = 'hello world'
s[0] = 'H'

# Python strings are immutable
# We can not change strings once they are made.
```

```python
s = 'hello world'
del s
print(s)
# it is already deleted, so we can't delete.
```

```python
#strings doesn't support item deletion because it is immutable, we can't change it.
s = 'hello world'
del s[-1:-5:2]
print(s)
```

```python
# Find the length of a given string without using the len() function

s = input('enter the string')

counter = 0

for i in s:
  counter += 1

print('length of string is',counter)
```

```python
## Find the total characters of a given string without whitespaces without using the len() function
a=(input("Enter string:")).strip().replace(" ","")
c=0

for i in a:
  c+=1
print("Total characters:",c)
```

```python
# Extract username from a given email.
# Eg if the email is nitish24singh@gmail.com
# then the username should be nitish24singh

s = input('enter the email: ')

pos = s.index('@') # finds the index of @
print(s[0:pos]) # slice string from start upto @

```

```python
# Count the frequency of a particular character in a provided string.
# Eg 'hello how are you' is the string, the frequency of h in this string is 2.

s = input('enter the email: ')
term = input('what would like to search for: ')

counter = 0
for i in s:
  if i == term:
    counter += 1

print('frequency is:',counter)
```

```python
# Write a program which can remove a particular character from a string.
s = input('enter the string: ')
term = input('what would like to remove: ')

result = ''

for i in s: # loops through s
  if i != term: #checks if the term is equal to s or not
    result = result + i # if not add to result

print(result)
```

```python
# Write a program that can check whether a given string is palindrome or not.
# abba
# malayalam

s = input('enter the string')
flag = True
for i in range(0,len(s)//2):
  if s[i] != s[len(s) - i -1]:
    flag = False
    print('Not a Palindrome')
    break

if flag:
  print('Palindrome')

```

```python
# Write a program to count the number of words in a string without split()

s = input('enter the string')
L = []
temp = ''
for i in s:

  if i != ' ':
    temp = temp + i
  else:
    L.append(temp)
    temp = ''

L.append(temp)
print(L)

```

```python
# Write a python program to convert a string to title case without using the title()
s = input('enter the string')

L = []
for i in s.split():
  L.append(i[0].upper() + i[1:].lower())

print(" ".join(L))
```

```python
# Write a program that can convert an integer to string.

number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result
  number = number//10

print(result)
print(type(result))
```
