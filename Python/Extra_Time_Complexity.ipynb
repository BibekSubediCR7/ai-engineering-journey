## Identify Big O notation.

```python
number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result
  number = number//10

print(result)


#345 loops runs 4 times, for 3450 it runs 5 times
# when increased by factor of 10, output time increase by 1.
#This is property of log. so O(logn) is time complexity.
```

```python
L = [1,2,3,4]

sum = 0
for i in L:
  sum = sum + i

product = 1
for i in L:
  product = product*i

print(sum,product)

#The loops are independent.
#so time complexity is O(n+n)=O(n)
```

```python
A = [1,2,3,4]
B = [5,6,7,8]
for i in A:
  for j in B:
    print(i,j)
#loops are dependent
#so = O(n^2)
```

```python
A = [1,2,3,4]
B = [5,6,7,8]

for i in A:
  for j in B:
    for k in range(1000000):
      print(i,j)

#all are nestes so O(n^2 * 1000000), since constant is taken out, it is O(n^2)

```

```python
L = [1,2,3,4,5]

for i in range(0,len(L)//2):
  other = len(L) - i -1
  temp = L[i]
  L[i] = L[other]
  L[other] = temp

print(L)

#loops runs n/2 times, there are 4 operations, so 4*n/2 =2n=O(n)
```

```python
n = 10
k = 0;
for i in range(n//2,n):
  for j in range(2,n,pow(2,j)):
        k = k + n / 2;

print(k)

#starts with n/2, and inner loop runs from 2 to n, taking 2 to power something, it is log.
# so O(nlogn)
```

```python
a = 10
b = 3

if b <= 0:
  print(-1)
div = a//b

print(a-div-b)

#O(1) because no loops, or recursion. They are just maths operations

# if no loop or recursion, always O(1)
```

```python
n = 345

sum = 0
while n>0:
  sum = sum + n%10
  n = n // 10

print(sum)


# it reduces by constant factor 10 every time, so it is O(logn)
```

```python
def fib(n):
  if n == 1 or n == 0:
    return 1
  else:
    return fib(n-1) + fib(n-2)


# this is recursion of fibonacci series

# it calls again and again the same function , if n tends to infinity it is exponential growth.

#count number of function calls.
#if n=3, fib(2),fib(1), for n=3, 5 function calls.
# for n=5, 15 function calls.

# it is like chain reaction, O(2^n)
```
