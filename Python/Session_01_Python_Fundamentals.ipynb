# 3. Variables

```python
# Static Vs Dynamic Typing
# Static Vs Dynamic Binding
# stylish declaration techniques

#Static Typing like in C/C++
# int a = 10, char nam="Bibek"

#Dynamic Typing in python
a=10
b="bibek"
# The variable auto detects the value that it holds; we don't have to define on our own.
print(a,b,sep="-")
```

```python
# In C/C++, we had to declare what each variable mean like int a =10, like that.
name = 'nitish'
print(name)

a = 5
b = 6

print(a + b)
a="Bibek"
print("My name is ",a)
#Python automatically detects what type of variable it holds depending if it hold number, float, char or strings, so it is dynamically typed.
```

```python
# Dynamic Typing
a = 5
# Static Typing
int a = 5
```

```python
# Dynamic Binding (Python)- The variable can hold any type of datatype in same program. Like a=2, a="bib"
a = 5
print(a)
a = 'nitish'
print(a)

# Static Binding (Java,C,C++)- The var datatype is fixed in the same program.
# int a = 5 (In C)
# If a=5 then we can't change the value of a to other datatype in same program.
```

# 7. Literals

```python
#Literals are the values given to variables,
a = 0b1010 #Binary Literals
b = 100 #Decimal Literal , 100 is the literal , b is variable.
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

#Float Literal
float_1 = 10.5
float_2 = 1.5e2 # 1.5 * 10^2
float_3 = 1.5e-3 # 1.5 * 10^-3

#Complex Literal
x = 3.14j

print(a, b, c, d)
print(float_1, float_2,float_3)
print(x, x.imag, x.real)
```

```python
# binary
x = 3.14j
print(x.imag)
```
