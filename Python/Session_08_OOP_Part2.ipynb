### Write OOP classes to handle the following scenarios:

- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line



```python
import math

class Point:

  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y

  def __str__(self):
    return f'<{self.x_cod},{self.y_cod}>'

  def euclidean_distance(self,other):
    return math.sqrt(((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2))

  def distance_from_origin(self):
    return math.sqrt((self.x_cod**2 + self.y_cod**2))
    # return self.euclidean_distance(Point(0,0))


class Line:

  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)

  def point_on_line(line,point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return "lies on the line"
    else:
      return "does not lie on the line"

  def shortest_distance(line,point):
    return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5
```

```python
p2=Point(2,4)
p3=Point(5,8)
Point.euclidean_distance(p2,p3) # same as p2.euclidean_distance(p3)
```

```python
l1 = Line(1,1,-2)
p1 = Point(1,10)
print(l1)
print(p1)

l1.shortest_distance(p1)
```

## Check if two lines intersect or not using Python oop

```python
class intersection:

  def __init__(self,a,b,c):
    self.a=a
    self.b=b
    self.c=c

  def __str__(self):
    return f'{self.a}x + {self.b}y + {self.c} = 0'

  def check_line(line1,line2):
    if (line1.a/line2.a)!=(line1.b/line2.b):
      return "The lines intersect!! and has solution"
    elif (line1.a/line2.a)==(line1.b/line2.b)!=(line1.c/line2.c):
      return "The lines are parallel and will never meet!!!, no solution."
    else:
      return "The lines are same lines, infinite solutions"
```

```python
l1=intersection(2,3,4)
l2=intersection(4,6,8)
print(l1)
print(l2)
print(l1.check_line(l2))

l3=intersection(2,3,1)
l4=intersection(4,6,8)
print(l3)
print(l4)
l3.check_line(l4)
```

### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
# object without a reference
class Person:

  def __init__(self):
    self.name = 'nitish'
    self.gender = 'male'

p = Person() # p is reference variable. p stores the address of object made in memory.
q = p
```

When creating object we use, obj_name=Classname(), but when we run Classname(), it runs and it also makes object which has address, we are storing that object adresss in obj_name variable.


```python
# Multiple ref
print(id(p))
print(id(q))
#both id is same which means they point to the same object
```

```python
# change attribute value with the help of 2nd object
```

```python
print(p.name)
print(q.name)
q.name = 'ankit'
print(q.name)
print(p.name)

# both p and q points to same memory address of object, so when changing one, also changes another.
```

### Pass by reference

```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person): # it takes object as parameter.
  print('Hi my name is',person.name,'and I am a',person.gender)
  p1 = Person('ankit','male')
  return p1 # this function is returning object p1.

p = Person('nitish','male') # object is created and stores in p
x = greet(p) #  that object is given to greet() as parameter
print(x.name)
print(x.gender)
```

```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  print(id(person))
  person.name = 'ankit'
  print(person.name)

p = Person('nitish','male')
print(id(p))
greet(p)
print(p.name)

# both are pointing to same object in memory so same memory address.
```

### Object ki mutability

```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  person.name = 'ankit'
  return person

p = Person('nitish','male')
print(id(p))
p1 = greet(p)
print(id(p1))

#even if there is change in attribute to a object, it still points to same address.
```

Mutable object: You can change its contents without creating a new object, so its identity/address stays the same.    
Immutable object: You cannot change the existing object. If you appear to "change" it, Python creates a new object and makes the variable refer to it.

### Static methods

##### Points to remember about static

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
- The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

```python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```
