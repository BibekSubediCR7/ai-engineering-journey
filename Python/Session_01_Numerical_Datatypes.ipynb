## Float

```python
# what are floating point numbers

x = 10.0
print(type(x))

y = -5.5
print(type(y))

z = 0.0
print(type(z))
```

```python
# precision --> numbers after the decimal point

a = 10.5 # lower precision
b = 10.513235594 # higher precision
```

```python
# inf can be represented as floats --> float()

x = float("inf")

print(x)
print(type(x))
```

```python
x = float("-inf")

print(x)
print(type(x))
```

```python
# operations

a = 10.5
b = 4.3

print(a + b)
print(a * b)
print(a / b)
print(a // b)
print(type(a // b))
```

```python
# adding int and float

a = 10.5

b = 2

print(type(a), type(b))

print(a + b)
```

```python
print(a // b)
print(type(a // b))
```

**float** - Higher order dtype as compared to int.

## Numerical representation in other formats

Base 10 - digits 0 - 9

Base 2 - Binary 0, 1

Base 8 - Octal 0 - 7

Base 16 - Hexadecimal - 0 - 9 A B C D E F

```python
# binary

# prefix 0b

x = "0b100"

print(x)

print(int(x, base=2))
```

```python
x = 4

print(x)

print(bin(x))

print(type(bin(x)))
```

```python
x = "0b100"

print(float(int(x, base=2)))
```

```python
# octal

# prefix 0o

x = "0o76"

print(int(x, base=8))
```

```python
oct(62)
```

```python
# hex

# prefix 0x

x = "0xAE23"

print(int(x, base=16))
```

```python
hex(44579)
```
