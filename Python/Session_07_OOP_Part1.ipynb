## Why Do We Need OOP?

OOP (Object-Oriented Programming) is used to **organize and manage complex programs** by grouping related data and functions into objects.

### Problem with Procedural Programming

As a program grows:
- Data and functions become scattered.
- Code becomes difficult to maintain and understand.
- The same code/structure may need to be repeated.
- Changes in one part can affect many other parts.

### What OOP Solves

OOP helps by:
- **Encapsulating** related data and functions together.
- **Reusing** code through classes and objects.
- **Organizing** large programs into manageable components.
- **Reducing** unnecessary repetition.
- Making code easier to **maintain, extend, and debug**.

> **In short:** Procedural programming focuses on *what steps to perform*, while OOP focuses on *what objects exist, what data they have, and what they can do*.

## Class

A **class** is a blueprint or template used to create objects. It defines the **attributes (data)** and **methods (behavior)** that its objects can have.  

Class can have data data/property or functions/behaviour or both.

For example:

class Car:  
    brand  
    color  
    start()  
    stop()  

Here, `Car` is a class and an actual car created from it is an object.  brand,color are data and start stop are functions.

### Python's Built-in Classes

Python itself is heavily based on classes. The built-in types we use every day are also **classes**.

For example:

- `int` → class for integers
- `float` → class for decimal numbers
- `str` → class for strings
- `list` → class for lists
- `tuple` → class for tuples
- `dict` → class for dictionaries
- `set` → class for sets

Someone implemented these classes with their associated methods and behavior.

For example:

x = [1, 2, 3]

Here, `list` is a class and `x` is an **object (instance)** of that class.

Because `x` is a list object, we can use methods defined by the `list` class:

x.append(4)
x.pop()
x.reverse()

Similarly:

name = "Bibek"

`name` is an object (instance) of the `str` class, so it has methods such as:

name.upper()
name.lower()
name.replace()

> **In short:** A class defines the structure and behavior, while an object is an instance created from that class. Python's `int`, `str`, `list`, `dict`, `tuple`, etc. are also classes.

## Functions vs Methods

Both **functions and methods are blocks of reusable code**, but the main difference is **where they belong and how they are called**.

### Function

A function is defined independently, outside a class.

def add(a, b):
    return a + b

add(2, 3)

Here, `add()` is a **function**.

### Method

A method is a function that is **defined inside a class** and is associated with an object/class.

```python
class Calculator:
    def add(self, a, b):
        return a + b

calc = Calculator()
calc.add(2, 3)
```



Here, `add()` is a **method** of the `Calculator` class.

`self` refers to the object that is calling the method.

### Python Example

numbers = [1, 2, 3]

numbers.append(4)

Here:
- `list` → class
- `numbers` → object
- `append()` → method of the `list` class

### In Short

| Function | Method |
|---|---|
| Defined independently | Defined inside a class |
| Called directly | Usually called through an object/class |
| `add(2, 3)` | `calc.add(2, 3)` |
| Not necessarily associated with an object | Associated with a class/object |

> **Every method is a function in terms of implementation, but a function becomes a method when it is defined as part of a class.**

## Magic Methods

**Magic methods** are special methods in Python that have **double underscores (`__`) at the beginning and end of their names**.

They are also called **dunder methods** (double underscore methods).

They are used to define how an object should behave with Python's built-in operations.

### Examples

```python

class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

    def __len__(self):
        return len(self.name)

person = Person("Bibek")

print(person)       # calls __str__()
len(person)         # calls __len__()
```




### Common Magic Methods

- `__init__()` → called when an object is created
- `__str__()` → defines what `print(object)` should display
- `__len__()` → defines what `len(object)` returns
- `__add__()` → defines behavior of `object1 + object2`
- `__eq__()` → defines behavior of `object1 == object2`
- `__lt__()` → defines behavior of `object1 < object2`

For example:

a + b

is internally related to:

a.__add__(b)

Similarly:

len(a)

is related to:

a.__len__()

> **In short:** Magic methods allow us to define how our custom objects behave with Python's built-in functions and operators.

Making Fraction datatype using class.

```python
class Fraction:

  # parameterized constructor
  def __init__(self,x,y): #This magic method runs the function when object is created,
    self.num = x
    self.den = y

  def __str__(self): #This magic method returns some value when we print the object.
    return '{}/{}'.format(self.num,self.den)

  def __add__(self,other): #This magic method returns some value when we use + operand b/w two or more objects.
    new_num = self.num*other.den + other.num*self.den
    new_den = self.den*other.den # new num and new den are temporary var used to store value so no self.var.

    return '{}/{}'.format(new_num,new_den)

  def __sub__(self,other): #This magic method returns  some value when we use - operanf b/w two or more objects.
    new_num = self.num*other.den - other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __mul__(self,other): # does multipication
    new_num = self.num*other.num
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __truediv__(self,other): # does tru division.
    new_num = self.num*other.den
    new_den = self.den*other.num

    return '{}/{}'.format(new_num,new_den)

  def convert_to_decimal(self):
    return self.num/self.den




```

```python
fr1 = Fraction(3,4)
fr2 = Fraction(1,2)
```

```python
fr1.convert_to_decimal()
# 3/4
```

```python
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
```
