### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
sal = float(input("Enter yearly salary in Lakhs: "))
#deductions first of all
deductions = (0.10 + 0.05 + 0.03) * sal
#taxable income now
taxable = sal - deductions
#for less than 5 lpa
if taxable <= 5:
    tax = 0
elif taxable <= 10:
    tax = (taxable - 5) * 0.10
elif taxable <= 20:
   tax = 5*0.1 + (taxable - 10) * 0.20
else:
   tax = 0.5 + 10*0.2 + (taxable - 20) * 0.30

annual_in_hand = (taxable - tax) * 100000
monthly_sal = annual_in_hand / 12

print("In-hand monthly salary for",sal,"Lakhs CTC is:",monthly_sal)
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Write code here
num=int(input("Enter a num:")) #5
fact=1
for i in range(1,num+1): #1,2,3,4,5
  fact=fact*i #1,2,6,24,120
print("Factorial of",num,"is",fact)
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash

```

`Output:`

```bash

```

````python
num = int(input("Enter a number N: "))

sum_val = 0
i = 1

while i <= num:
    if i % 5 != 0: # If i is not divisible by 5
        if sum_val + i > 300:# check if sum+i is greater than 300, if yes skips the if loop and terminates the while loop.
            break
        sum_val += i #if sum val+i is false , i is added to sum_val.
    i += 1 # i increments

print("Final sum:", sum_val)
````

````python
#Full version(AI Assisted Code)
import math

x, y = 0, 0

while True:
    move = input("Enter direction and steps (or ! to stop): ").strip()

    if move == "!":
        break

    direction, steps = move.split()
    steps = int(steps)

    if direction == "UP":
        y += steps
    elif direction == "DOWN":
        y -= steps
    elif direction == "LEFT":
        x -= steps
    elif direction == "RIGHT":
        x += steps

# Calculate distance from origin (0,0)
distance = math.sqrt(x**2 + y**2)

# Round to nearest integer
print(round(distance))
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here
num = int(input("Enter a number: "))

temp = num
sum_val = 0

while temp!=0:
    digit = temp % 10
    sum_val += digit**3
    temp = temp // 10

if sum_val == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
# Write code here
import math

H = int(input("Enter Hour: "))
M = int(input("Enter Minute: "))
if H == 12:
    H = 0

hour_angle = (H * 30) + (M * 0.5)
minute_angle = M * 6

angle = abs(hour_angle - minute_angle)

final_angle = min(angle, 360 - angle)

print(math.floor(final_angle))
````
