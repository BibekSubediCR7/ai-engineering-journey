## Operators in Python

- Arithmetic Operators
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators

```python
# Arithmetric Operators - Normal arithmetic operators
print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2) # integer divison, gives in integer.

print(5%2) # remainder

print(5**2) #5 to the power 2
print(71//3)
```

```python
# Relational Operators - Comparing 2 or more numbers
print(4>5)

print(4<5)

print(4>=4)

print(4<=4)

print(4==4)

print(4!=4)
print(True >= False) #1>=0, True
```

```python
# Logical Operators - and , or and not
print(1 and 0) # 1 if both are true otherwise false.

print(1 or 0) #1 ,0 if both are false else true

print(not 1)
```

```python
# Bitwise Operators - special operators which operators in binary values

# bitwise and
print(2 & 3) # convert 2 and 3 in binary digits , and apply and operation
# 2 is      010
# 3 is      110
#apply and  010 which is 2, thats why it shows 2.

# bitwise or
print(2 | 3) # same but uses or in this.

# bitwise xor - false if both are same , otherwise true
print(2 ^ 3)

#not of three - changes 1 to 0 and 0 to 1
print(~3)

#bitwise left shift operator
print(4 >> 2)

#bitwise right shift operator.
print(5 << 2)
```

```python
# Assignment Operators

# =
# a = 2

#a is var and 2 is literal
a = 2 # a equals to 2

# a = a % 2
a %= 2

# a++ post increment
# ++a # pre increment
# This is not allowed , post and pre increment
print(a)
```

```python
# Membership Operators - checks if something is present or not

# in/not in

print('D' not in 'Delhi')

print(1 in [2,3,4,5,6])
print("Bib" in ("Bib,564"))
```

### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
# Code Used in the session
curr_pop = 10000

for i in range(2026,2016,-1):
  print(i,curr_pop)
  curr_pop=curr_pop/1.1


```

#### Explanation :



To calculate the population for each year with a 10% increase, you can use a simpler equation based on the previous year's population. Let's assume the population of the previous year is represented by variable x.

The equation can be written as:

> Current Year Population = x * 1.1

In this equation, the current year's population is equal to the previous year's population multiplied by 1.1, representing a 10% increase.

To find the population of the previous year (x), we can rearrange the equation as follows:

> x = Current Year Population / 1.1

Using this simplified equation, if you have the current year's population (e.g., 10,000), you can divide it by 1.1 to calculate the population of the previous year.

This equation allows you to calculate the population for each year, assuming you know the population of the current year and want to find the population of the previous year.
